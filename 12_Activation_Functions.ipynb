## What is an Activation Function?
- Without it: network = just a linear model (no matter how many layers!)
- With it: network can learn **non-linear** patterns
- Applied after linear step: `output = activation(W*X + b)`

## All Activation Functions
| Function | Range | Best For |
|---|---|---|
| Sigmoid | (0,1) | Output binary classification |
| Tanh | (-1,1) | RNNs hidden layers |
| ReLU | [0,inf) | Hidden layers (most common) |
| Leaky ReLU | (-inf,inf) | Fix dying ReLU |
| ELU | (-alpha,inf) | Smoother ReLU |
| Softmax | (0,1) sum=1 | Multi-class output |
| GELU | smooth | Transformers/BERT |

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-6, 6, 300)

sigmoid    = lambda x: 1/(1+np.exp(-x))
tanh       = lambda x: np.tanh(x)
relu       = lambda x: np.maximum(0, x)
leaky_relu = lambda x: np.where(x>0, x, 0.01*x)
elu        = lambda x: np.where(x>0, x, np.exp(x)-1)
gelu       = lambda x: 0.5*x*(1+np.tanh(np.sqrt(2/np.pi)*(x+0.044715*x**3)))

funcs = [('Sigmoid',sigmoid(x),'blue'),('Tanh',tanh(x),'green'),
         ('ReLU',relu(x),'red'),('Leaky ReLU',leaky_relu(x),'purple'),
         ('ELU',elu(x),'orange'),('GELU',gelu(x),'brown')]

fig, axes = plt.subplots(2,3,figsize=(15,8))
for ax,(name,y,col) in zip(axes.flatten(),funcs):
    ax.plot(x,y,color=col,lw=3)
    ax.axhline(0,color='k',lw=0.5); ax.axvline(0,color='k',lw=0.5)
    ax.set_title(name,fontsize=13,fontweight='bold')
    ax.set_xlabel('x'); ax.set_ylabel('f(x)'); ax.grid(True,alpha=0.3)
plt.tight_layout(); plt.show()
```

```python
# Sigmoid — vanishing gradient demo
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-8,8,300)
sig = 1/(1+np.exp(-x))
sig_d = sig*(1-sig)

fig,(a1,a2)=plt.subplots(1,2,figsize=(10,4))
a1.plot(x,sig,'b-',lw=3); a1.set_title('Sigmoid: f(x)=1/(1+e^-x)'); a1.grid(True,alpha=0.3)
a2.plot(x,sig_d,'r-',lw=3); a2.set_title("Sigmoid Derivative — Max=0.25 (Vanishing Gradient!)"); a2.grid(True,alpha=0.3)
plt.tight_layout(); plt.show()

print("Sigmoid range: (0,1) — good for probability output")
print("Problem: max gradient = 0.25 → shrinks each layer → VANISHING GRADIENT")
print("Use only in: Output layer for binary classification")
```

```python
# ReLU vs Leaky ReLU — Dying ReLU Demo
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-4,4,300)
relu       = np.maximum(0,x)
leaky      = np.where(x>0,x,0.01*x)
relu_d     = np.where(x>0,1.0,0.0)
leaky_d    = np.where(x>0,1.0,0.01)

fig,(a1,a2)=plt.subplots(1,2,figsize=(12,4))
a1.plot(x,relu,'r-',lw=3,label='ReLU'); a1.plot(x,leaky,'b--',lw=3,label='Leaky ReLU (a=0.01)')
a1.set_title('ReLU vs Leaky ReLU'); a1.legend(); a1.grid(True,alpha=0.3)
a2.plot(x,relu_d,'r-',lw=3,label="ReLU' (0 or 1)"); a2.plot(x,leaky_d,'b--',lw=3,label="Leaky ReLU'")
a2.set_title('Derivatives — ReLU flat at 0 = Dying ReLU Problem!'); a2.legend(); a2.grid(True,alpha=0.3)
plt.tight_layout(); plt.show()

print("Dying ReLU: if neuron always gets x<0, gradient=0, never learns")
print("Fix: Leaky ReLU gives small gradient (0.01) even for x<0")
```

```python
# Softmax — multi-class output
import numpy as np

def softmax(z):
    e = np.exp(z - np.max(z))  # subtract max for stability
    return e / e.sum()

z = np.array([2.0, 1.0, 0.1])  # 3-class logits
p = softmax(z)
print("Logits (raw scores)  :", z)
print("Softmax probabilities:", p.round(4))
print("Sum                  :", p.sum().round(6), "(always = 1.0)")
print(f"Class 0: {p[0]*100:.1f}% | Class 1: {p[1]*100:.1f}% | Class 2: {p[2]*100:.1f}%")
print("Prediction: Class", np.argmax(p))
```

## Interview Questions

**Q: Why is ReLU better than Sigmoid for hidden layers?**
> ReLU gradient = 1 for x>0 (no vanishing gradient). Sigmoid max gradient = 0.25, shrinks layer by layer in deep networks. ReLU also faster to compute.

**Q: What is the Dying ReLU problem?**
> If neuron always gets x<0, ReLU always outputs 0, gradient=0, weights never update. Neuron is permanently dead. Fix: Leaky ReLU (slope 0.01 for x<0), ELU, He initialization.

**Q: Why Softmax in multi-class output?**
> Converts raw scores to probabilities summing to 1. Each class gets a probability. Natural with Categorical Cross-Entropy loss.

**Q: Where is GELU used?**
> BERT, GPT, and transformer models. Smoother than ReLU, works better with attention mechanisms.
