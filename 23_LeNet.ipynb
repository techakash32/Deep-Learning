## 📌 What is LeNet? (Simple English)
- **First successful CNN** — designed to recognize handwritten digits (MNIST)
- Created by **Yann LeCun** in 1998
- Very simple by today's standards — only **7 layers**
- Used **Average Pooling** (before MaxPool became popular)
- Used **Sigmoid/Tanh** activations (before ReLU)
- Proved that CNNs work — started the deep learning revolution

## 🔑 LeNet-5 Architecture
```
Input (32×32×1)
→ Conv(6, 5×5) → AvgPool(2×2)       # Feature maps: 14×14×6
→ Conv(16,5×5) → AvgPool(2×2)       # Feature maps: 5×5×16
→ Flatten (400)
→ Dense(120) → Dense(84) → Dense(10)
```

## 🧱 LeNet vs Modern CNNs
| Feature | LeNet (1998) | Modern CNN (2024) |
|---|---|---|
| Layers | 7 | 100s+ |
| Activation | Sigmoid/Tanh | ReLU |
| Pooling | Average | Max |
| Params | ~60K | Millions |
| Dataset | MNIST (32×32) | ImageNet (224×224) |

````python
import tensorflow as tf
import numpy as np

# ── LeNet-5 Architecture ──
def build_lenet():
    model = tf.keras.Sequential([
        # Layer 1: Conv + AvgPool
        tf.keras.layers.Conv2D(6, (5,5), activation='tanh', input_shape=(32,32,1)),
        tf.keras.layers.AveragePooling2D(2,2),

        # Layer 2: Conv + AvgPool
        tf.keras.layers.Conv2D(16,(5,5), activation='tanh'),
        tf.keras.layers.AveragePooling2D(2,2),

        # Classifier
        tf.keras.layers.Flatten(),
        tf.keras.layers.Dense(120, activation='tanh'),
        tf.keras.layers.Dense(84,  activation='tanh'),
        tf.keras.layers.Dense(10,  activation='softmax'),
    ], name='LeNet-5')
    return model

lenet = build_lenet()
lenet.summary()
print(f"\nTotal params: {lenet.count_params():,}  (≈60K — tiny!)")
````

````python
# Train LeNet on MNIST
(x_train, y_train),(x_test, y_test) = tf.keras.datasets.mnist.load_data()
# LeNet uses 32×32 input — resize from 28×28
x_train = np.pad(x_train, ((0,0),(2,2),(2,2))).reshape(-1,32,32,1).astype('float32')/255.0
x_test  = np.pad(x_test,  ((0,0),(2,2),(2,2))).reshape(-1,32,32,1).astype('float32')/255.0

lenet.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
lenet.fit(x_train, y_train, epochs=5, batch_size=128, validation_split=0.1, verbose=1)
_, acc = lenet.evaluate(x_test, y_test, verbose=0)
print(f"\nLeNet MNIST Accuracy: {acc:.4f}")
````

## 🗣️ Interview Q&A

**Q: What is LeNet and who created it?**
> LeNet-5 was created by Yann LeCun in 1998. It was the first successful CNN used for handwritten digit recognition. It proved that convolutional networks could learn from images.

**Q: What are the key components of LeNet?**
> Input (32×32) → Conv(6 filters, 5×5) → AvgPool → Conv(16 filters, 5×5) → AvgPool → Flatten → Dense(120) → Dense(84) → Dense(10, softmax)

**Q: What is the historical significance of LeNet?**
> It was the first practical demonstration that CNNs work. It was used by banks to read cheque digits. It inspired all future deep learning architectures.

**Q: What limitations did LeNet have?**
> Used sigmoid/tanh (slow, vanishing gradients), avg pooling (not max), needed 32×32 input, only 7 layers — too shallow for complex tasks like ImageNet.
