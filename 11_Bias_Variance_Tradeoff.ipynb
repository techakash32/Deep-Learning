## 📌 What is Bias? (Simple English)
- **Bias** = how much the model's predictions are **systematically wrong**
- High Bias = model makes same type of mistake consistently
- High Bias = **Underfitting** — too simple to learn patterns
- Example: fitting a straight line to a curved dataset

## 📌 What is Variance? (Simple English)
- **Variance** = how much the model **changes** when trained on different data
- High Variance = model is very sensitive to training data
- High Variance = **Overfitting** — memorizes noise
- Example: wavy polynomial that follows every data point

## 📌 The Tradeoff
- You can't reduce both at the same time (usually)
- Simpler model → High Bias, Low Variance
- Complex model → Low Bias, High Variance
- **Goal: find the sweet spot!**

## 🔑 Formula
```
Total Error = Bias² + Variance + Irreducible Noise
```

## 📊 Comparison Table
| | Bias | Variance | Error on Train | Error on Test |
|---|---|---|---|---|
| **Underfitting** | High ❌ | Low | High | High |
| **Overfitting** | Low | High ❌ | Low | High |
| **Good Model** | Low ✅ | Low ✅ | Low | Low |

````python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import Pipeline
from sklearn.model_selection import cross_val_score

np.random.seed(42)

def true_func(x):
    return np.sin(x)

# Generate training data (small — to show variance)
n_train = 15
X_train = np.random.uniform(0, 2*np.pi, n_train).reshape(-1,1)
y_train = true_func(X_train.ravel()) + np.random.normal(0, 0.2, n_train)

X_plot = np.linspace(0, 2*np.pi, 200).reshape(-1,1)
y_true = true_func(X_plot.ravel())

# Show predictions from different model complexities
degrees = [1, 4, 15]
titles  = ['High Bias (degree=1)
Underfitting',
           'Good Balance (degree=4)',
           'High Variance (degree=15)
Overfitting']
colors  = ['red', 'green', 'blue']

fig, axes = plt.subplots(1, 3, figsize=(15, 4))
for ax, deg, title, col in zip(axes, degrees, titles, colors):
    pipe = Pipeline([('poly', PolynomialFeatures(deg)), ('lr', LinearRegression())])
    pipe.fit(X_train, y_train)
    y_pred = pipe.predict(X_plot)
    
    ax.scatter(X_train, y_train, alpha=0.7, label='Training data', zorder=5)
    ax.plot(X_plot, y_true, 'k--', lw=2, label='True function')
    ax.plot(X_plot, y_pred, color=col, lw=2, label=f'Model (deg={deg})')
    ax.set_title(title); ax.legend(fontsize=7)
    ax.set_ylim(-2.5, 2.5); ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
````

````python
# Bias-Variance vs Model Complexity curve
degrees = range(1, 16)
train_errors, val_errors = [], []

X_full = np.random.uniform(0, 2*np.pi, 200).reshape(-1,1)
y_full = true_func(X_full.ravel()) + np.random.normal(0, 0.2, 200)

for deg in degrees:
    pipe = Pipeline([('poly', PolynomialFeatures(deg)), ('lr', LinearRegression())])
    cv_train = -cross_val_score(pipe, X_full, y_full, cv=5, scoring='neg_mean_squared_error')
    val_errors.append(cv_train.mean())

    pipe.fit(X_full, y_full)
    y_pred_tr = pipe.predict(X_full)
    train_errors.append(np.mean((y_full - y_pred_tr)**2))

plt.figure(figsize=(10, 5))
plt.plot(list(degrees), train_errors, 'b-o', label='Training Error (Bias²)', lw=2)
plt.plot(list(degrees), val_errors,   'r-o', label='Validation Error (Bias²+Variance)', lw=2)
plt.axvline(x=4, color='green', linestyle='--', label='Sweet Spot', lw=2)
plt.xlabel('Model Complexity (Polynomial Degree)')
plt.ylabel('Mean Squared Error')
plt.title('Bias-Variance Tradeoff')
plt.legend(); plt.grid(True, alpha=0.3)
plt.ylim(0, 1)
plt.show()
print("Left of green line → High Bias (Underfit)")
print("Right of green line → High Variance (Overfit)")
print("Green line → Sweet Spot ✅")
````

## 🗣️ Interview Questions & Answers

**Q: What is the bias-variance tradeoff?**
> Simple models have high bias (underfitting) + low variance. Complex models have low bias + high variance (overfitting). Total error = Bias² + Variance + Noise. The goal is to find the sweet spot that minimizes total test error.

**Q: How do you know if your model has high bias or high variance?**
> High bias: both train AND test error are high.
> High variance: train error is low but test error is high (big gap).
> Compare train vs val/test performance to diagnose.

**Q: How does regularization help the bias-variance tradeoff?**
> Regularization (Dropout, L1/L2) adds constraints that reduce variance (prevents overfitting) at a slight cost of increased bias. It's a deliberate bias-variance trade toward better generalization.

**Q: What is irreducible error?**
> Noise in the data that no model can remove — no matter how perfect. Example: measurement errors, randomness in real world. Total Error = Bias² + Variance + Irreducible Error.
