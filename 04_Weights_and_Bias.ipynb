## 📌 What are Weights? (Simple English)
- Weights = how much **importance** to give each input
- High weight → that feature matters a lot
- Zero weight → feature is ignored
- Negative weight → feature has opposite effect
- Weights are **learned during training** (via backprop + optimizer)

## 📌 What is Bias? (Simple English)
- Bias = a **constant shift** added to the neuron output
- Without bias: decision boundary must pass through origin (too restrictive)
- Bias lets the model fit data **not centered at zero**
- Same role as **intercept (b)** in `y = mx + b`

## 🔑 Full Neuron Formula
```
z = (w₁×x₁) + (w₂×x₂) + ... + (wₙ×xₙ) + b
output = activation(z)
```

## 🧠 Easy Analogy
| Neural Network | Real Life |
|---|---|
| Weights | How much you trust each person's opinion |
| Bias | Your default opinion before hearing anyone |
| Input | Each person's recommendation |
| Output | Your final decision |

````python
import numpy as np

# Single Neuron Calculation — manual
x = np.array([2.0, 3.0, 1.0])   # 3 input features
w = np.array([0.5, -0.3, 0.8])  # weights
b = 0.1                          # bias

z = np.dot(x, w) + b
print("Inputs  :", x)
print("Weights :", w)
print("Bias    :", b)
print(f"z = {x[0]}×{w[0]} + {x[1]}×{w[1]} + {x[2]}×{w[2]} + {b} = {z:.4f}")
print(f"After ReLU: {max(0, z):.4f}")

print("\n💡 Weights control DIRECTION, Bias controls SHIFT")
````

````python
import tensorflow as tf
import numpy as np

model = tf.keras.Sequential([
    tf.keras.layers.Dense(3, activation='relu', input_shape=(4,)),
    tf.keras.layers.Dense(1, activation='sigmoid')
])

print("=== BEFORE TRAINING ===")
for i, layer in enumerate(model.layers):
    w, b = layer.get_weights()
    print(f"Layer {i+1} | W shape:{w.shape} | b:{b.round(4)}")
    print(f"  Weights (first row): {w[0].round(4)}")
````

````python
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

X, y = make_classification(n_samples=500, n_features=4, random_state=42)
X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.2)
sc = StandardScaler()
X_tr = sc.fit_transform(X_tr); X_te = sc.transform(X_te)

model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
model.fit(X_tr, y_tr, epochs=30, verbose=0)

print("=== AFTER TRAINING ===")
for i, layer in enumerate(model.layers):
    w, b = layer.get_weights()
    print(f"Layer {i+1} | b:{b.round(4)} | W(first row):{w[0].round(4)}")
print("\n✅ Weights changed — the model LEARNED from data!")
````

````python
# Weight Initialization methods
import tensorflow as tf

inits = {
    'Random Normal' : tf.keras.initializers.RandomNormal(mean=0, stddev=0.05),
    'Glorot Uniform': tf.keras.initializers.GlorotUniform(),   # default for sigmoid/tanh
    'He Normal'     : tf.keras.initializers.HeNormal(),        # best for ReLU
    'Zeros'         : tf.keras.initializers.Zeros(),           # ❌ BAD — symmetry problem
}

print(f"{'Init Method':20s} | mean    | std    | range")
print("-"*60)
for name, init in inits.items():
    w = init(shape=(5,5)).numpy()
    print(f"{name:20s} | {w.mean():+.4f} | {w.std():.4f} | [{w.min():.3f}, {w.max():.3f}]")
````

## 🗣️ Interview Questions & Answers

**Q: What happens if all weights are initialized to zero?**
> All neurons compute the same output → same gradients → same weight updates → network never learns. Called the **symmetry problem**. Always use random initialization.

**Q: What is Xavier/Glorot initialization?**
> Initializes weights with variance = `2/(fan_in + fan_out)`. Prevents vanishing/exploding gradients. Best for sigmoid/tanh.

**Q: What is He initialization?**
> Variance = `2/fan_in`. Designed for **ReLU** activations. Standard choice in modern deep networks.

**Q: Why do we need bias?**
> Without bias, the activation function is fixed at origin. Bias shifts it, giving the model flexibility to fit any pattern regardless of where data is centered.
