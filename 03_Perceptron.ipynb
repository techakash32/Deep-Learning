## 📌 What is a Perceptron? (Simple English)
- Perceptron = **simplest possible neural network** — just 1 neuron
- Binary classifier: output is **0 or 1** (uses step function)
- Takes inputs → multiplies by weights → adds bias → applies step function
- Invented by **Frank Rosenblatt in 1958**
- A single perceptron ≈ **Logistic Regression** conceptually
- ❌ Can only solve **linearly separable** problems (fails on XOR)

## 🔑 Perceptron Math
```
z      = w₁x₁ + w₂x₂ + ... + wₙxₙ + b
output = 1  if z ≥ 0
         0  if z < 0
```

## 🧱 Perceptron → MLP
| Type | Description |
|---|---|
| **Single Perceptron** | 1 neuron, linear boundary only |
| **Multi-Layer Perceptron (MLP)** | Many neurons in layers |
| **Deep Network** | MLP with many hidden layers |

````python
import numpy as np

# Build Perceptron from scratch
class Perceptron:
    def __init__(self, lr=0.1, epochs=100):
        self.lr = lr
        self.epochs = epochs
        self.w = None
        self.b = 0

    def step(self, z):
        return np.where(z >= 0, 1, 0)

    def fit(self, X, y):
        self.w = np.zeros(X.shape[1])
        for _ in range(self.epochs):
            for i in range(len(X)):
                pred  = self.step(np.dot(X[i], self.w) + self.b)
                error = y[i] - pred
                # Update rule: w = w + lr * error * x
                self.w += self.lr * error * X[i]
                self.b += self.lr * error

    def predict(self, X):
        return self.step(np.dot(X, self.w) + self.b)

# --- AND Gate (linearly separable → Perceptron CAN learn)
X = np.array([[0,0],[0,1],[1,0],[1,1]])
y_and = np.array([0, 0, 0, 1])

p = Perceptron()
p.fit(X, y_and)
print("AND Gate:")
print("  Predictions:", p.predict(X))
print("  Expected   :", y_and, "✅")
````

````python
# --- XOR Gate (NOT linearly separable → Perceptron FAILS)
y_xor = np.array([0, 1, 1, 0])

p_xor = Perceptron()
p_xor.fit(X, y_xor)
print("XOR Gate:")
print("  Predictions:", p_xor.predict(X))
print("  Expected   :", y_xor, "❌ Cannot solve XOR!")
print()
print("Solution → Multi-Layer Perceptron (add hidden layers)")

# MLP solves XOR
import tensorflow as tf
mlp = tf.keras.Sequential([
    tf.keras.layers.Dense(4, activation='relu', input_shape=(2,)),
    tf.keras.layers.Dense(1, activation='sigmoid')
])
mlp.compile(optimizer='adam', loss='binary_crossentropy')
mlp.fit(X, y_xor, epochs=500, verbose=0)
preds = (mlp.predict(X, verbose=0) > 0.5).astype(int).flatten()
print("MLP XOR predictions:", preds, "✅ Solved!")
````

## 🗣️ Interview Questions & Answers

**Q: What is a Perceptron?**
> A single artificial neuron with binary output (0 or 1). It computes a weighted sum of inputs + bias and applies a step function. The simplest neural network.

**Q: What is the biggest limitation of a single Perceptron?**
> It can only learn **linearly separable** patterns. XOR problem cannot be solved. Solution: stack multiple perceptrons into a Multi-Layer Perceptron (MLP).

**Q: What is the Perceptron learning rule?**
> `w_new = w_old + learning_rate × (actual - predicted) × x`
> Weights only update when a mistake is made.

**Q: Perceptron vs Logistic Regression?**
> Both linear classifiers. Perceptron → hard step (0/1). LR → sigmoid (probability). LR is differentiable → works better with gradient descent and gives probabilities.
