## 📌 What is a Loss Function? (Simple English)
- Loss = **how wrong your model is** on current predictions
- Lower loss = better model = closer to correct answers
- Goal of training = **minimize** the loss
- Also called: cost function, objective function

## 🔑 Common Loss Functions
| Loss Function | Task | Formula |
|---|---|---|
| **MSE** | Regression | mean((y - ŷ)²) |
| **MAE** | Regression | mean(|y - ŷ|) |
| **Binary Crossentropy** | Binary Classification | -[y·log(ŷ) + (1-y)·log(1-ŷ)] |
| **Categorical Crossentropy** | Multi-class (one-hot) | -Σ y·log(ŷ) |
| **Sparse Categorical CE** | Multi-class (integers) | same, labels as integers |
| **Huber Loss** | Regression + outliers | MSE for small, MAE for large errors |

## 🎯 Quick Decision Guide
| Task | Label Format | Use This Loss |
|---|---|---|
| Regression | continuous number | MSE or MAE |
| Binary Classification | 0 or 1 | Binary Crossentropy |
| Multi-class | one-hot encoded | Categorical Crossentropy |
| Multi-class | integer labels | Sparse Categorical Crossentropy |

```python
import numpy as np

y_true = np.array([1.0, 0.0, 1.0, 1.0, 0.0])
y_pred = np.array([0.9, 0.1, 0.8, 0.3, 0.6])   # model predictions

# 1. MSE
mse = np.mean((y_true - y_pred)**2)
print(f"MSE  : {mse:.4f}  ← squares errors, penalizes large mistakes heavily")

# 2. MAE
mae = np.mean(np.abs(y_true - y_pred))
print(f"MAE  : {mae:.4f}  ← absolute errors, treats all mistakes equally")

# 3. Binary Crossentropy
eps = 1e-9
bce = -np.mean(y_true*np.log(y_pred+eps) + (1-y_true)*np.log(1-y_pred+eps))
print(f"BCE  : {bce:.4f}  ← for binary classification")

# Perfect prediction → BCE ≈ 0
y_perf = np.array([0.999, 0.001, 0.999, 0.999, 0.001])
bce_p = -np.mean(y_true*np.log(y_perf+eps) + (1-y_true)*np.log(1-y_perf+eps))
print(f"\nBCE (perfect predictions): {bce_p:.6f} ≈ 0")
print(f"BCE (our predictions)     : {bce:.4f}  (higher → worse)")
```

```python
import tensorflow as tf
import numpy as np

y_true_tf = tf.constant([1.0, 0.0, 1.0, 1.0, 0.0])
y_pred_tf = tf.constant([0.9, 0.1, 0.8, 0.3, 0.6])

losses = {
    'MSE' : tf.keras.losses.MeanSquaredError(),
    'MAE' : tf.keras.losses.MeanAbsoluteError(),
    'BCE' : tf.keras.losses.BinaryCrossentropy(),
}
for name, fn in losses.items():
    print(f"TF {name}: {fn(y_true_tf, y_pred_tf).numpy():.4f}")
```

```python
import matplotlib.pyplot as plt
import numpy as np

errors = np.linspace(-3, 3, 200)

plt.figure(figsize=(10, 4))
plt.plot(errors, errors**2,         label='MSE (squared)',       color='red',    lw=2)
plt.plot(errors, np.abs(errors),    label='MAE (absolute)',      color='blue',   lw=2)
huber = np.where(np.abs(errors)<=1, 0.5*errors**2, np.abs(errors)-0.5)
plt.plot(errors, huber,             label='Huber (best of both)',color='green',  lw=2)

plt.title('Loss Functions Comparison')
plt.xlabel('Error (y - ŷ)')
plt.ylabel('Loss Value')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

print("MSE: large errors hurt MUCH more (squared)")
print("MAE: all errors treated equally (robust to outliers)")
print("Huber: MAE for outliers + MSE for small errors ← best of both")
```

## 🗣️ Interview Questions & Answers

**Q: Why Binary Crossentropy instead of MSE for classification?**
> BCE is derived from Maximum Likelihood Estimation for probabilities. It gives large gradients when predictions are very wrong, making learning faster. MSE gives small gradients near 0/1 boundaries.

**Q: MSE vs MAE — when to use which?**
> MSE → when large errors should be penalized more (outliers are important to catch)
> MAE → when data has outliers you want to ignore (more robust)

**Q: What is Sparse Categorical Crossentropy?**
> Same math as Categorical CE but labels are integers (0, 1, 2...) not one-hot. More memory efficient with many classes.

**Q: What does it mean when loss stops decreasing?**
> Model has converged (good) or is stuck in a local minimum. Try different learning rate, optimizer, or architecture.
