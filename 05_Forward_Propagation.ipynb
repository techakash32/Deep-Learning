## 📌 What is Forward Propagation? (Simple English)
- Forward propagation = data travels **input → output** through all layers
- Each layer: compute `z = W·X + b`, apply activation → pass to next layer
- Result = the model's **prediction (ŷ)**
- Then we compute **loss** = how wrong the prediction is
- Think of it like an **assembly line** — each station processes and passes forward

## 🔑 Step-by-Step
```
Step 1: X enters Input Layer
Step 2: z₁ = W₁ · X + b₁       (linear combo)
Step 3: a₁ = activation(z₁)     (apply activation)
Step 4: z₂ = W₂ · a₁ + b₂      (next layer)
Step 5: a₂ = activation(z₂)     (final output = ŷ)
Step 6: Compute Loss(ŷ, y)
```

## 🔑 Notation Table
| Symbol | Meaning |
|---|---|
| X | Input features |
| W | Weight matrix |
| b | Bias vector |
| z | Pre-activation (linear output) |
| a | Post-activation (non-linear output) |
| ŷ | Final prediction |

````python
import numpy as np

np.random.seed(42)

# 3 samples, 4 features
X = np.array([[1.0, 2.0, 0.5, 1.5],
              [0.3, 1.1, 0.8, 0.2],
              [2.0, 0.5, 1.2, 0.9]])

# Layer 1: 4 → 3 neurons
W1 = np.random.randn(4, 3) * 0.1
b1 = np.zeros((1, 3))

# Layer 2: 3 → 1 neuron
W2 = np.random.randn(3, 1) * 0.1
b2 = np.zeros((1, 1))

def relu(z):    return np.maximum(0, z)
def sigmoid(z): return 1 / (1 + np.exp(-z))

# === FORWARD PASS ===
print(f"Input shape: {X.shape}")

z1 = X.dot(W1) + b1
print(f"\nLayer 1 — z1 (pre-activation)  shape: {z1.shape}")

a1 = relu(z1)
print(f"Layer 1 — a1 (after ReLU)       shape: {a1.shape}")
print(f"a1 values:\n{a1.round(4)}")

z2 = a1.dot(W2) + b2
a2 = sigmoid(z2)
print(f"\nFinal output (predictions):\n{a2.round(4)}")
print("Predicted classes:", (a2 > 0.5).astype(int).flatten())
````

````python
# Same thing in TensorFlow — compare outputs
import tensorflow as tf

model = tf.keras.Sequential([
    tf.keras.layers.Dense(3, activation='relu', input_shape=(4,)),
    tf.keras.layers.Dense(1, activation='sigmoid')
])

# Set same weights as our manual calculation
model.layers[0].set_weights([W1, b1[0]])
model.layers[1].set_weights([W2, b2[0]])

pred_tf = model.predict(X, verbose=0)
print("Manual forward pass output:", a2.flatten().round(4))
print("TensorFlow output          :", pred_tf.flatten().round(4))
print("\n✅ Identical — TF does the same math internally!")
````

## 🗣️ Interview Questions & Answers

**Q: What is forward propagation?**
> Passing input data layer-by-layer through the network (computing z = W·X + b, then activation), to produce a final prediction ŷ.

**Q: What is the difference between z and a?**
> z = pre-activation (linear: W·X + b). a = post-activation (non-linear: activation(z)). Non-linearity is what makes deep learning powerful.

**Q: Why is forward prop needed before backward prop?**
> Backprop needs the activations (a values) from each layer stored during forward pass. These are used to compute gradients. No forward pass → no backprop possible.

**Q: What happens during forward propagation vs backpropagation?**
> Forward: compute prediction left-to-right (input → output)
> Backward: compute gradients right-to-left (output → input) to update weights
