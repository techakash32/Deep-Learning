## 📌 What is Gradient Descent? (Simple English)
- Gradient Descent = the **optimization algorithm** that trains neural networks
- It **minimizes the loss** by updating weights in the direction that reduces loss
- "Gradient" = slope of the loss curve (how steep is it?)
- "Descent" = go downhill toward the minimum
- Like a hiker finding the lowest point of a valley by always stepping downhill

## 🔑 Update Rule
```
w_new = w_old  -  learning_rate × gradient
                      (η)           (dL/dw)
```
- If gradient is **positive** → loss increases with w → decrease w
- If gradient is **negative** → loss decreases with w → increase w

## 🧱 Types of Gradient Descent
| Type | Data per step | Speed | Stability |
|---|---|---|---|
| **Batch GD** | All data | Slow | Very stable |
| **Stochastic GD (SGD)** | 1 sample | Fast | Noisy |
| **Mini-batch GD** | Small batch (32/64) | Balanced ✅ | Good |

````python
import numpy as np
import matplotlib.pyplot as plt

# Manual Gradient Descent — Linear Regression example (from your class code!)
np.random.seed(13)
X = np.array([-1.5, -0.5, 0.5, 1.5])
y = np.array([-80, -20, 60, 120])

m = 78.35  # slope (fixed)
b = 0      # intercept (we optimize this)
lr = 0.1

print("=== Gradient Descent: Finding best intercept (b) ===")
print(f"Starting b = {b}")
print()

history = [b]
for step in range(4):
    loss_slope = -2 * np.sum(y - m*X - b)     # dLoss/db
    step_size  = loss_slope * lr
    b_new      = b - step_size
    
    loss_before = np.mean((y - m*X - b)**2)
    loss_after  = np.mean((y - m*X - b_new)**2)
    
    print(f"Step {step+1}: b={b:.4f} | gradient={loss_slope:.2f} | step={step_size:.2f} | b_new={b_new:.4f}")
    print(f"         Loss: {loss_before:.2f} → {loss_after:.2f}")
    
    b = b_new
    history.append(b)

print(f"\nFinal b = {b:.4f}")
````

````python
# Visualize gradient descent convergence
losses = [np.mean((y - m*X - bi)**2) for bi in history]

plt.figure(figsize=(12, 4))

plt.subplot(1,2,1)
plt.plot(range(len(history)), history, 'bo-', markersize=8)
plt.xlabel('Step'); plt.ylabel('Intercept (b)')
plt.title('Intercept converging'); plt.grid(True, alpha=0.3)

plt.subplot(1,2,2)
plt.plot(range(len(losses)), losses, 'ro-', markersize=8)
plt.xlabel('Step'); plt.ylabel('Loss (MSE)')
plt.title('Loss decreasing ✅'); plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
````

````python
# Effect of different learning rates
b_values = np.linspace(-50, 150, 300)
losses_curve = [np.mean((y - m*X - b)**2) for b in b_values]

fig, axes = plt.subplots(1, 3, figsize=(14, 4))
learning_rates = [0.001, 0.1, 0.5]
labels = ['Too Small (slow)', 'Just Right ✅', 'Too Large (diverges)']

for ax, lr_val, label in zip(axes, learning_rates, labels):
    b_cur = 0.0
    path = [b_cur]
    for _ in range(10):
        grad = -2*np.sum(y - m*X - b_cur)
        b_cur = b_cur - lr_val * grad
        path.append(b_cur)
        if abs(b_cur) > 300: break  # diverged
    
    ax.plot(b_values, losses_curve, 'k-', lw=2)
    for i, bp in enumerate(path[:-1]):
        if abs(bp) < 150:
            lv = np.mean((y - m*X - bp)**2)
            ax.plot(bp, lv, 'ro', markersize=8)
    
    ax.set_title(f"lr={lr_val}\n{label}")
    ax.set_xlabel('b'); ax.set_ylabel('Loss')
    ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
````

## 🗣️ Interview Questions & Answers

**Q: What is gradient descent?**
> An optimization algorithm that iteratively updates model weights in the direction of steepest loss decrease, using: `w = w - lr × gradient`.

**Q: What is the learning rate and why is it important?**
> LR controls how big each step is. Too small → very slow training. Too large → overshoots minimum, diverges. Typically 0.001–0.01.

**Q: What is a local minimum vs global minimum?**
> Global minimum = lowest loss possible. Local minimum = lowest in a small region but not globally. GD can get stuck in local minima. Solutions: momentum, Adam, random restarts.

**Q: What is the difference between Batch, SGD, and Mini-batch GD?**
> Batch: use all data per step (stable, slow). SGD: 1 sample per step (fast, noisy). Mini-batch: small batch (32/64) — best of both — used in practice.
