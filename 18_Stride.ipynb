## 📌 What is Stride? (Simple English)
- Stride = **how many pixels the filter jumps** at each step
- Stride=1 → move 1 pixel at a time (default, detailed)
- Stride=2 → skip every other pixel (faster, smaller output)
- Larger stride → **smaller output**, fewer computations
- Stride is sometimes used **instead of pooling** to reduce spatial size

## 🔑 Output Size Formula
```
Output = floor((Input - Kernel + 2×Padding) / Stride) + 1

Example: Input=28, Kernel=3, Padding=0, Stride=2
→ floor((28 - 3 + 0) / 2) + 1 = floor(12.5) + 1 = 13
```

## 🧱 Stride vs Pooling
| | Stride > 1 | Pooling |
|---|---|---|
| Reduces size | ✅ | ✅ |
| Learnable | ✅ (filter) | ❌ (fixed operation) |
| Info loss | Less | More |
| Modern CNNs | Preferred ✅ | Also used |

````python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

image = np.arange(1, 50, dtype=np.float32).reshape(7,7)
img_tf = image.reshape(1,7,7,1)
ones_kernel = np.ones((3,3,1,1), dtype=np.float32)

# Different strides
results = {}
for s in [1, 2, 3]:
    out = tf.nn.conv2d(img_tf, ones_kernel, strides=s, padding='VALID').numpy().squeeze()
    results[s] = out

print("Input shape: 7×7")
for s, out in results.items():
    print(f"Stride={s} → Output: {out.shape}")

fig, axes = plt.subplots(1, 4, figsize=(14,4))
axes[0].imshow(image, cmap='YlOrRd'); axes[0].set_title('Input 7×7')
for i,(s,out) in enumerate(results.items(),1):
    axes[i].imshow(out, cmap='YlOrRd')
    axes[i].set_title(f'Stride={s} → {out.shape}')
for ax in axes: ax.axis('off')
plt.tight_layout(); plt.show()
````

````python
# Stride effect in Keras CNN
import tensorflow as tf

# stride=1 (default)
m1 = tf.keras.Sequential([tf.keras.layers.Conv2D(32,(3,3),strides=1,padding='same',input_shape=(28,28,1))])
# stride=2 (halves size)
m2 = tf.keras.Sequential([tf.keras.layers.Conv2D(32,(3,3),strides=2,padding='same',input_shape=(28,28,1))])
# stride=3
m3 = tf.keras.Sequential([tf.keras.layers.Conv2D(32,(3,3),strides=3,padding='same',input_shape=(28,28,1))])

for name,m in [("stride=1",m1),("stride=2",m2),("stride=3",m3)]:
    m.build()
    out = m.layers[0].output_shape
    print(f"{name} → output: {out}  | params: {m.count_params()}")
print("\nStride=2 halves width and height → 4× fewer spatial values!")
````

## 🗣️ Interview Q&A

**Q: What is stride in CNN?**
> The number of pixels the filter moves at each step. Stride=1 → move 1px (dense coverage). Stride=2 → jump 2px (reduces output size by half).

**Q: How does stride reduce computation?**
> Larger stride → fewer positions the filter visits → smaller feature map → fewer computations in the next layer.

**Q: Stride vs Pooling for downsampling?**
> Both reduce spatial size. Strided convolution is **learnable** (the network decides how to downsample). Pooling is fixed (max or average). Modern architectures (ResNet, EfficientNet) prefer strided convolutions.

**Q: Output size formula with stride?**
> `floor((Input - Kernel + 2×Padding) / Stride) + 1`
